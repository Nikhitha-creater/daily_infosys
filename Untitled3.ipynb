```python
import nltk
from nltk.stem import WordNetLemmatizer

# Download WordNet dictionary (required for lemmatization)
nltk.download('wordnet')
nltk.download('omw-1.4')  # recommended for complete word support

lemmatizer = WordNetLemmatizer()

print("stones →", lemmatizer.lemmatize("stones"))     # plural → singular
print("speaking →", lemmatizer.lemmatize("speaking", pos='v'))  # verb → base form
print("bedroom →", lemmatizer.lemmatize("bedroom"))   # compound words usually unchanged
print("cats →", lemmatizer.lemmatize("cats"))         # plural → singular
```

```python
from nltk.stem import SnowballStemmer

# Create English stemmer
snow = SnowballStemmer("english")

# Words to stem
words = ["running", "studies", "playing", "caring", "happily", "cats"]

# Print stems
for w in words:
    print(w, "→", snow.stem(w))
```
